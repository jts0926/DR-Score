# Single bilateral radiograph inference
Detect and crop both knees and generate one DR Score per knee using the included synthetic example. The resulting scores demonstrate the software workflow only and are not clinical results.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
```

```python
from drscore.inference import infer_bilateral_image

IMAGE_PATH = ROOT / 'examples' / 'synthetic_bilateral_knee_example.png'
OUTPUT_DIR = ROOT / 'outputs' / 'single_case'

result = infer_bilateral_image(
    IMAGE_PATH,
    OUTPUT_DIR,
    config_path=ROOT / 'configs' / 'final_model.yaml',
    patient_right_on_image_left=True,
)
result
```
